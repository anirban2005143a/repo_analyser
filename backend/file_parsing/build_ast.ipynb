```python
import json
import tree_sitter_python as tspython
from tree_sitter import Language, Parser

# Initialize language and parser
PY_LANGUAGE = Language(tspython.language())
parser = Parser(PY_LANGUAGE)

with open("nodes.py", "rb") as f:
    source_code = f.read()

tree = parser.parse(source_code)

def node_to_dict(node, source_code):
    # Extract the source code text corresponding to this node
    node_text = source_code[node.start_byte:node.end_byte].decode("utf-8", errors="ignore")[:100]
    
    return {
        "type": node.type,
        "start_point": list(node.start_point),  # Convert tuple to list for JSON compatibility
        "end_point": list(node.end_point),
        "start_byte": node.start_byte,
        "end_byte": node.end_byte,
        "text": node_text,
        "children": [node_to_dict(child, source_code) for child in node.children]
    }

# Convert the root node and its children into a dictionary
tree_dict = node_to_dict(tree.root_node, source_code)

# Format the dictionary into a pretty-printed JSON string
json_output = json.dumps(tree_dict, indent=2)
print(json_output)

# Optional: Save the JSON output to a file for easier inspection
with open("parsed_tree.json", "w", encoding="utf-8") as f:
    f.write(json_output)
```

```python
import os
import json
from tree_sitter import Parser
from tree_sitter_language_pack import get_language

def parse_code_to_json(file_path):
    # 1. Identify language via file extension
    ext = os.path.splitext(file_path)[1].lower()
    lang_name = EXTENSION_MAP.get(ext)
    
    if not lang_name:
        raise ValueError(f"No parser mapping found for extension '{ext}'")
    
    print(f"File: {file_path} --> Language Pack Parser: '{lang_name}'")

    # 2. Retrieve language and setup the modern parser
    language_obj = get_language(lang_name)
    parser = Parser(language_obj)
    
    with open(file_path, "rb") as f:
        source_code = f.read()

    # 3. Parse code into AST
    tree = parser.parse(source_code)

    # 4. Recursive AST-to-JSON formatter
    def node_to_dict(node):
        node_text = source_code[node.start_byte:node.end_byte].decode("utf-8", errors="ignore")[:100]
        return {
            "type": node.type,
            "start_point": list(node.start_point),
            "end_point": list(node.end_point),
            "start_byte": node.start_byte,
            "end_byte": node.end_byte,
            "text": node_text ,  # Suppress token content on structural parents for a tidy JSON tree
            "children": [node_to_dict(child) for child in node.children]
        }

    # 5. Output JSON
    tree_dict = node_to_dict(tree.root_node)
    return json.dumps(tree_dict, indent=2)

# --- Example Usage ---
if __name__ == "__main__":
    file_to_parse = "./test.pl"  # Swap with any file (.js, .rs, .go, .java, .cpp, etc.)
    
    json_output = parse_code_to_json(file_to_parse)
    print(json_output)

    # Save to file
    with open("parsed_tree.json", "w", encoding="utf-8") as f:
        f.write(json_output)
```
